```python
# Q7: Sum of weights
cars_asia = data[data["origin"] == "Asia"].copy()
cars_asia = cars_asia[["vehicle_weight", "model_year"]].iloc[:7]
X = cars_asia.values
XTX = X.T @ X
inv_XTX = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = inv_XTX @ X.T @ y
print(sum(w))
```
